```python
from astropy.coordinates import EarthLocation, AltAz, get_sun
from astropy.time import Time, TimeDelta
from astropy.units import deg
import numpy as np

# from astroplan import Observer
```

```python
def get_sun_altaz(location, time):
    frame = AltAz(location=location, obstime=time)
    sun = get_sun(time).transform_to(frame)
    return (sun.alt.value, sun.az.value) # (altitude, azimuth)
```

```python
# setting the location.
lat, lon = -15.783507525862756*deg, -47.91526919090653*deg # praça do cruzeiro
location = EarthLocation(lat=lat, lon=lon)
```

```python
# handling time the right way
from datetime import datetime
from zoneinfo import ZoneInfo

utc = ZoneInfo('UTC')
time = datetime(2025, 5, 26, hour=16).replace(tzinfo=utc)
```

```python
current_tz = ZoneInfo('America/Sao_Paulo')
sp_time = time.astimezone(tz=current_tz)

print(time.__repr__(), 'UTC')
print(sp_time.__repr__(), 'America/Sao_Paulo' )
print(sp_time.replace(tzinfo=utc).__repr__(), '(replaced to UTC)')
```

```python
astro_time = Time(time)
utc_astro_time  = Time(time, scale='utc')

sp_astro_time = Time(sp_time)
utc_sp_astro_time = Time(sp_time, scale = 'utc')

astro_time, utc_astro_time, sp_time.__repr__(), sp_astro_time, utc_sp_astro_time
```
