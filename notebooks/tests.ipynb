```python
# Testing whether datetime objects are mutable or immutable. They are immutable.
from datetime import datetime

def teste(date: datetime):
    print('inside teste', date)
    date: int = 1
    print('inside teste, after change', date)

date = datetime(year=2026,month=1,day=1,hour=6,minute=18)
print('before call teste', date)
result = teste(date)
print('after call teste', date)

# testing identity.
date2 = date.replace(year=2025) # returns a new object
print('date2 is date:', date2 is date)
```
