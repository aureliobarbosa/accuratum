```python
from astropy.coordinates import EarthLocation, AltAz, get_sun
from astropy.time import Time, TimeDelta
from astropy.units import deg
import numpy as np

from astroplan import Observer
```

```python
def get_sun_altaz(location, time):
    frame = AltAz(location=location, obstime=time)
    sun = get_sun(time).transform_to(frame)    
    return (sun.alt.value, sun.az.value) # (altitude, azimuth)
```

```python
# setting the location.
lat, lon = -15.783507525862756*deg, -47.91526919090653*deg # praça do cruzeiro
location = EarthLocation(lat=lat, lon=lon)
```

```python
# handling time the right way
from datetime import datetime
from zoneinfo import ZoneInfo

current_tz = ZoneInfo('America/Sao_Paulo')
time = datetime(2025, 5, 26, hour=16).replace(tzinfo=current_tz)
astro_time = Time(time, scale='local')

print('datetime aware: ', time, current_tz)
print('astro_time:     ', astro_time, 'utc' )
```

```python
time2 = astro_time.to_datetime().replace(tzinfo=current_tz)
print(time2)
```

```python
# get altitude and azimuth coordinates
coords = get_sun_altaz(location=location,time=astro_time)
coords
```

```python
obs = Observer(location=location)
sun_rise = obs.sun_rise_time(astro_time,which='next').datetime.replace(tzinfo=current_tz)
sun_set = obs.sun_set_time(astro_time,which='next').datetime.replace(tzinfo=current_tz)
print('sunset:', sun_set)
print('sunrise:', sun_rise)
```
